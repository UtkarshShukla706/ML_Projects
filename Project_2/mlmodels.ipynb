```python
df.drop_duplicates()
df.isnull().sum()
df.dropna(subset=['embarked'],inplace=True)
print(df.isnull().sum())
```

```python
df['age'].skew()
#hence the skewness of the value is between -0.5 to 0.5 then it is the acceptable hence the null values are filled by the mean of the data 
df.fillna(df['age'].mean(), inplace=True)
```

```python
#now then go to the another step of the processing of the data for the models

df_clean=df.copy()
```

```python
df_clean['alone']=df_clean['alone'].astype(int)
df_clean
```

```python
#now i am go for the label encoding of the sex columns

df_clean['sex'].value_counts()
```

```python
#now we have to perform the label encoding in the sex columns and one hot encoding in the embarked column

from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df_clean['sex']=le.fit_transform(df_clean['sex'])
```

```python
df_clean
```

```python
#for the one hot encoding
```

```python
df_clean= pd.get_dummies(df_clean, columns=['embarked'],drop_first=True).astype(int)
```

```python
df_clean.head(100)
df_clean_sample=df_clean.copy()
```

```python
from sklearn.preprocessing import StandardScaler
cols=['age','fare']
scaler=StandardScaler()
df_clean[cols]=scaler.fit_transform(df_clean[cols])
```

```python
df_clean
```

```python
#now hence the model is ready now i have to apply the different types of the model in it 

x=df_clean.drop('survived',axis=1)
y=df_clean['survived']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)

#now what we have do we just seperate our data into train test and split format we split the data into the training data and the testing data
```

```python
#after the training and testing of the data 

from sklearn.linear_model import LinearRegression

model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
#now for the testing of the linear regression model we need to find the r2 score mean sqaure error and rmse absolute r2 score

from sklearn.metrics import mean_squared_error,r2_score,root_mean_squared_error
r2=r2_score(y_test,y_pred)

print("R2 Score is :",r2)
print("MSE Score is :",mean_squared_error(y_test,y_pred))
print("RMSE Score is :",root_mean_squared_error(y_test,y_pred))
n=x_test.shape[0]
p=x_test.shape[1]

adjusted_r2=1-(1-r2)*(n-1)/(n-p-1)
print("Adjusted R2 Score is: ",adjusted_r2)
```

```python
#now we are going to use the different model that is Logistic Regression 

from sklearn.linear_model import LogisticRegression

lr=LogisticRegression()
lr.fit(x_train,y_train)

```

```python
#we are going to check the predicted value of the model

y_pred=lr.predict(x_test)
```

```python
#we have to check the accuracy of our model so in this we are going to use the Confusion matrix and calculate the confusion matrix etc

#it means we have to calculate both the accuracy precision and the recall and Accuracy Score 

from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

#now we check accordingly 

accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
#now lets go and check another model that is KNN

#so here the standard scaler is already applied 

from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=5)

knn.fit(x_train,y_train)

```

```python
y_pred=knn.predict(x_test)
```

```python
#for testing the model we need to do the same thing 

accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
#now what we are going to do we are going to apply the Ensemble learning in our model 

from sklearn.ensemble import StackingClassifier

baselearners=[
    {'dt',DecisionTreeClassifier(random_state=42)},
    {'svc',SVC(kernel='rbf',random_state=42,probability=True)},
    {'lr',LogisticRegression(max_iter=1000)},
    {'knn',KNeighborsClassifier(n_neighbors=5)}
]
```

```python
meta_learners=KNeighborsClassifier(n_neighbors=5)
```

```python
stacking_clf=StackingClassifier(
    estimators=baselearners,
    final_estimator=meta_learners,
    cv=5
)

stacking_clf.fit(x_train,y_train)
```

```python
y_pred_stacking=stacking_clf.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_stacking)
```

```python
print(classification_report(y_test,y_pred_stacking))
```
